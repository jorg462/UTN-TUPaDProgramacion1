#### Una aclaración importante acerca de la generación de números aleatorios: la semilla

Cuando se trabaja con programas que tienen una componente estocástica, durante la etapa de desarrollo, es conveniente utilizar siempre los mismos números aleatorios para poder comparar los resultados de una ejecución con la siguiente. 

En caso de que ocurran errores, lo ideal es que estos ocurran siempre bajo las mismas condiciones o en el mismo lugar, esto me asegura de que el uso de números aleatorios no provoque que los errores ocurran unas veces si y otras no. 

Para esto, se puede utilizar el método `seed()` pasandole un número, ese número será la semilla, el inicializador con el cual el generador de números aleatorios comienza a dar una secuencia de números. Si utilizo la misma semilla el generador me devolverá siempre la misma secuencia de números aleatorios. **IMPORTANTE**: la semilla se deberia colocar una sola vez en su código, cambiar la semilla dentro de una misma ejecución de su código puede provocar efectos no deseados. Por defecto, el generador de numeros aleatorios utiliza la hora del sistema como semilla. 

Vea el funcionamiento del siguiente ejemplo:

```python
import random

print("Genero 5 numeros con la semilla 10:")
random.seed(10)
for i in range(5):
    print(random.random()) 

print("\nGenero 5 numeros nuevos con la misma semilla:")
random.seed(10)
for i in range(5):
    print(random.random()) 
    
print("\nGenero 5 numeros con una semilla nueva:")
random.seed(11)
for i in range(5):
    print(random.random()) 
```
